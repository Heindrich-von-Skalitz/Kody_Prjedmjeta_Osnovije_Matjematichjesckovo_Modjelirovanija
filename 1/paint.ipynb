```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
def importTen(filename,dim):
    data = np.loadtxt(filename)
    return data.reshape(dim)
```

```python
Nx = 2048
Nt= 32768
```

```python
Xm = importTen("Xm.txt",(Nx,Nt))
Tm = importTen("Tm.txt",(Nx,Nt))
u = importTen("u.txt",(Nx,Nt))
```

```python
plt.figure(figsize=(10, 6))
plt.pcolormesh(Xm, shading='auto', cmap='viridis')
plt.colorbar(label='Z value')
plt.xlabel('T')
plt.ylabel('X')
plt.title('Plot of X')
plt.show()

plt.figure(figsize=(10, 6))
plt.pcolormesh(Tm, shading='auto', cmap='viridis')
plt.colorbar(label='Z value')
plt.xlabel('T')
plt.ylabel('X')
plt.title('Plot of T')
plt.show()

plt.figure(figsize=(10, 6))
plt.pcolormesh(Xm, Tm, u, shading='auto', cmap='viridis')
plt.colorbar(label='Z value')
plt.xlabel('T')
plt.ylabel('X')
plt.title('Plot of u')
plt.show()
```
